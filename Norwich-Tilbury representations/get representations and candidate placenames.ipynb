```python
import PyPDF2
PDFFile = open("EN020027-000941-EN020027 Reps links for Exam Library.pdf",'rb')

PDF = PyPDF2.PdfReader(PDFFile)
pages = len(PDF.pages)
key = '/Annots'
uri = '/URI'
ank = '/A'

representations_list = []


# for page in range(pages):
for pagenum, page in enumerate(PDF.pages):
    # print("Current Page: {}".format(page))
    pageSliced = page
    pageObject = pageSliced.get_object()
    if key in pageObject.keys():
        ann = pageObject[key]
        for a in ann:
            u = a.get_object()
            if uri in u[ank].keys():
                print(u[ank][uri])
                representations_list.append(u[ank][uri])
```

```python
startstr = '<div class="pins-rte">\\n\\t\\t\\t\\t<p>'
endstr = "</p>\\n\\t\\t\\t</div>"
problems = {}
n = 0

for url in representations_list:
    if url[-9:-5] == '1000':
        id = url[-9:]
        response = requests.get(url)
        if response.status_code == 200:
            content = str(response.content)
            n1 = content.find(startstr)
            n2 = content.find(endstr,n1)

            if n1>-1 and n2 > -1:
                with open(f"all_representations/{id}.txt", "w") as fl:
                    fl.write(content[n1+len(startstr):n2])

                    print(n, id)
                    n+=1
        else:
            problems[id] = response.status_code
            continue

# Took 5h21 without multiprocessing, checking 4008 urls and successfully retrieving from 3747.
# The other ones must have the wrong format as they are not in 'problems'. Yes, it's all the docs with format like 'https://nsip-documents.planninginspectorate.gov.uk/published-documents/EN020027-000552-2%20Norwich%20to%20Tilbury%20Examination%20Library%20%20PDF%20.pdf' and they seem to be dead links for the most part.
```

```python
def get_candidate_placenames(text):

    excluded_words = set(["Norwich", "Tilbury", "Suffolk", "Norfolk", "Essex"]).union(set(["The", "I", "We", "Also", "This", "A", "In", "Moreover", "Redacted", "There", "It", "They", "To"]))

    split_lines = ",".join(text.splitlines())
    splitbypunctuation = re.split(r'[.,;:/<>()?!\\]', split_lines)

    candidate_place_names = []
    # for phrase in splitbypunctuation:
    for n,phrase in enumerate(splitbypunctuation):
        # print(n, phrase)
        mid_candidate_placename = False
        words = phrase.strip().split(" ")
        current_candidate = ""
        nwords = 0
        for m, word in enumerate(words):
            # print(word)
            if len(word) > 0 and word[0].isupper() and word not in excluded_words:
                current_candidate += f" {word}"
                nwords += 1
                mid_candidate_placename = True
                # print(word, current_candidate)
            else:
                if mid_candidate_placename:
                    if nwords == 1:
                        if current_candidate.strip().upper() not in dictionary_words:
                            candidate_place_names.append(current_candidate.strip())
                    else:
                        candidate_place_names.append(current_candidate.strip())
                    current_candidate = ""
                    nwords = 0
                mid_candidate_placename = False
        if current_candidate != "":
            if current_candidate.strip().upper() not in dictionary_words:
                candidate_place_names.append(current_candidate.strip())

    return set(candidate_place_names).difference(excluded_words)
```

```python
directory = Path('all_representations')

candidate_places_dict = {}

for n,file in enumerate(list(directory.iterdir())):
    if file.is_file():
        id = file.name[:9]
        with open(file, "r") as f:
            text = f.read()
            candidates = get_candidate_placenames(text)
            if len(candidates):
                candidate_places_dict[id] = list(candidates)
    print(n, end = ",")

candidate_places_dict
```
